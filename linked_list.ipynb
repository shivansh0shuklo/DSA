```python
#singly node
class singlynode:
    def __init__(self,val,next =None):
        self.val = val
        self.next = next
    def __str__(self):
        return str(self.val)
```

```python
head = singlynode(1)
a = singlynode(3)
b = singlynode(4)
c = singlynode(7)
head.next = a
a.next = b
b.next = c
print(head)
```

```python
#traverse the list - 0(n)
curr = head
while curr:
    print(curr)
    curr = curr.next
```

```python
#display linked list  - 0(n)
def display(head):
    curr = head
    elements = []
    while curr:
        elements.append(str(curr.val))
        curr = curr.next
    print(" --> ".join(elements))#['a','b','c'] 'a' --> 'b' --> 'c'
display(head)
```

```python
def search(head,val):
    curr = head
    while curr:
        if val == curr.val:
            return True
        curr = curr.next
    return False
search(head,7)
```

```python
#doubly linked list 
class doubly:
    def __init__(self,val,next=None,prev=None):
        self.val = val
        self.next = next
        self.prev = prev
    def __str__(self):
        return str(self.val)
    
```

```python
head = tail = doubly(1)
print(tail)
```

```python
#display - o(n)
def display(head):
    curr = head
    elements = []
    while curr:
        elements.append(str(curr.val))
        curr = curr.next
    print(" <-> ".join(elements))
display(head)
```

```python
def insert_at_begining(head,tail,val):
    new_node = doubly(val,next=head)
    head.prev = new_node
    return new_node,tail
head,tail = insert_at_begining(head,tail,3)
display(head)
```

```python
def insert_at_end(head,tail,val):
    new_node = doubly(val,prev=tail)
    tail.next = new_node
    return head,new_node
head,tail = insert_at_end(head,tail,7)
display(head)
```

```python
head = tail = doubly(1)
head,tail = insert_at_begining(head, tail,3)
head, tail = insert_at_end(head,tail,7)
display(head)
```
