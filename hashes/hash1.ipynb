```python
#HASHSETS
s = set()
print(s)
```

```python
#add item into the set - O(1)
s.add(1)
s.add(2)
s.add(3)
print(s)
```

```python
#lookup if item in the set - O(1)
if 1 not in s:
    print(True)
```

```python
#removing from the set o(1)
s.remove(3)
print(s)
```

```python
string = 'aaaaaaaaaaaaaaaaaabbbbbbcccccccccccddddddeeeeee'
sett=  set(string)#set construction is O(s) - s is the length of the string
print(sett)
```

```python
#loop over the tems in set - O(n)
for x in s:
    print(x)
```

```python
#hashmaps == dictionary(in pyhton)
d ={"greg":1,'steave':2,'rob':3}
print(d)
```

```python
##adding key:val pair in dictionary : O(1)
d['rome'] = 4
print(d)
```

```python
#check of the presence of key key is in the dctionary

if 'greg' in d:
    print(True)
```

```python
#check the value corresponding to a key in the dictionary:o(1)
print('greg' in d)
print(d['greg'])
```

```python
#loop over the key;val pairs of the dictionary : o(n)
for key,val in d.items():
    print(f'key {key} : val {val}')
```
